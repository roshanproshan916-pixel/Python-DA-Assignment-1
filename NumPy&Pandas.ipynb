## PART-1 NumPy Array

## 1. Create a 1D NumPy Array
Create a NumPy array named temperatures_w1 containing the
daily average temperatures for Week 1.

```python
# Import NumPy library
import numpy as np

# Create a 1D array for Week 1 temperatures
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Display the array
print("Week 1 Temperatures:")
print(temperatures_w1)
```

## 2. Inspection and Properties
Inspect the shape, data type, and number of elements
in the temperatures_w1 array.

```python
# Display the shape of the array
print("Shape:", temperatures_w1.shape)

# Display the data type
print("Data Type:", temperatures_w1.dtype)

# Display the number of elements
print("Number of Elements:", temperatures_w1.size)
```

## 3. Array Operations
Convert Celsius temperatures into Fahrenheit and find
the maximum, minimum, and mean temperatures.

```python
# Convert Celsius to Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32

# Find maximum temperature
print("Maximum Temperature:", np.max(temperatures_w1))

# Find minimum temperature
print("Minimum Temperature:", np.min(temperatures_w1))

# Find mean temperature
print("Mean Temperature:", np.mean(temperatures_w1))

# Display Fahrenheit temperatures
print("Fahrenheit Temperatures:", fahrenheit)
```


## 4. Array Slicing and Indexing
Extract the temperatures for the first three days,
last two days, and middle three days of the week.

```python
# Extract first three days
print("First Three Days:", temperatures_w1[:3])

# Extract weekend temperatures (last two days)
print("Weekend Temperatures:", temperatures_w1[-2:])

# Extract middle three days
print("Middle Three Days:", temperatures_w1[2:5])
```

## 5. Create a 2D NumPy Array
Create a 2D array named temperatures containing
the temperatures of Week 1 and Week 2.

```python
# Create a 2D array for two weeks
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

# Display the 2D array
print("Temperatures of Two Weeks:")
print(temperatures)
```

## 6. Inspect and Slice the 2D Array
Display the shape, data type, and total number of elements.
Extract temperatures for each week and weekends of both weeks.

```python
# Display shape
print("Shape:", temperatures.shape)

# Display data type
print("Data Type:", temperatures.dtype)

# Display total number of elements
print("Total Elements:", temperatures.size)

# Extract Week 1 temperatures
print("Week 1:", temperatures[0])

# Extract Week 2 temperatures
print("Week 2:", temperatures[1])

# Extract weekend temperatures for both weeks
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

## PART-2 PANDAS

## 1. Creating Pandas Series
Create a Pandas Series named marks with the given marks
and custom indices.

```python
# Import Pandas library
import pandas as pd

# Create a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# Display the Series
print(marks)
```

## 2. Indexing and Slicing
Access marks using integer index position, loc, iloc
and apply boolean masking to filter marks greater than 90.

```python
# Access the mark of 1st rank student using integer position
print("1st Rank Mark:", marks.iloc[0])

# Retrieve marks of top 3 ranks using loc
print("\nTop 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve the mark of 3rd rank student using iloc
print("\n3rd Rank Mark:", marks.iloc[2])

# Filter marks greater than 90 using boolean mask
print("\nMarks Greater Than 90:")
print(marks[marks > 90])
```

## 3. Manipulating Series
Modify the 1st rank mark, remove the last rank,
and calculate CGPA by dividing each mark by 10.

```python
# Modify the 1st rank student's mark to 100
marks.loc['Rank1'] = 100

# Display modified Series
print("Modified Marks:")
print(marks)

# Remove the last rank from the Series
marks = marks.drop('Rank5')

# Display Series after removing last rank
print("\nAfter Removing Last Rank:")
print(marks)

# Calculate CGPA by dividing each mark by 10
cgpa = marks / 10

# Display CGPA
print("\nCGPA:")
print(cgpa)
```

## Part 3 – Pandas DataFrame

## 1. Creating Pandas DataFrame
Create a DataFrame named transactions using the given
TransactionID, ProductCategory, Region and Amount data.

```python
# Create a DataFrame using dictionary
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Display the DataFrame
print(transactions)
```

## 2. Data Exploration
Explore the DataFrame using head, tail, shape, column names,
data types, column selection, filtering, value counts,
unique values and groupby operations.

```python
# Display the complete DataFrame
print("Transactions DataFrame:")
print(transactions)
```

```python
# Display first five rows
print("First Five Rows:")
print(transactions.head())

# Display last five rows
print("\nLast Five Rows:")
print(transactions.tail())
```

```python
# Display DataFrame shape
print("Shape:", transactions.shape)

# Display column names
print("\nColumn Names:")
print(transactions.columns)

# Display data types
print("\nData Types:")
print(transactions.dtypes)

# Display basic information
print("\nDataFrame Information:")
transactions.info()
```

```python
# Display ProductCategory and Amount columns
print("Product Category and Amount:")
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve the last three columns using iloc
print("Last Three Columns:")
print(transactions.iloc[:, -3:])
```

```python
# Filter rows where Region is North and Amount is greater than 200
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

# Display filtered data
print("Filtered Transactions:")
print(filtered_data)
```

```python
# Find the count of each ProductCategory
print("Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())
```

```python
# Find unique values in Region column
print("Unique Regions:")
print(transactions['Region'].unique())
```

```python
# Calculate mean Amount for each Region
print("Average Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

## 3. Manipulating the DataFrame
Modify the Amount for TransactionID 102, add a Discount column,
remove TransactionID 109 and delete the Discount column.

```python
# Modify Amount for TransactionID 102 to 165
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

# Display updated DataFrame
print("After Modifying Amount:")
print(transactions)
```

```python
# Calculate 10% discount of Amount
transactions['Discount'] = transactions['Amount'] * 0.10

# Display updated DataFrame
print("After Adding Discount Column:")
print(transactions)
```

```python
# Remove the row with TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

# Display updated DataFrame
print("After Removing TransactionID 109:")
print(transactions)
```

```python
# Delete the Discount column
transactions = transactions.drop(columns=['Discount'])

# Display final DataFrame
print("Final Transactions DataFrame:")
print(transactions)
```
